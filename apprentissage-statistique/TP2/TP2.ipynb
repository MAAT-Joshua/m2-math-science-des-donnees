# Exercice 2 

On veut écrire une fonction python qui permette de déterminer le meilleur modèle entre g1 et g2 d'après la fonction de perte indicatrice (qui compte le nombre d'erreur dans le modèle).

```python
import numpy as np

def risque_empirique(N):
    m = len(N)
    total = 0
    erreurs = 0

    for i in range(m):
        for j in range(m):
            total = total + N[i][j]
            if i != j:
                erreurs = erreurs + N[i][j]
    return erreurs / total

def meilleur_modele(N1, N2):
    r1 = risque_empirique(N1)
    r2 = risque_empirique(N2)

    if r1 < r2:
        return "g1"
    elif r2 < r1:
        return "g2"
    else:
        return "égalité"
```

On teste :

```python
import numpy as np
g1 = np.array([[44, 0, 0], [5, 62, 1], [1, 8, 54]])
g2 = np.array([[44, 4, 5], [2, 64, 3], [4, 2, 47]])
meilleur_modele(g1,g2)
```

On retrouve bien $g_1$ meilleur modèle (on a fait que vérifié ce qu'on a fait à la main à l'exercice 1)
