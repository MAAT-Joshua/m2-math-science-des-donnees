# C – Maximum à posteriori 

1) Solution maximum a posteriori (MAP) à
la question, en utilisant le cadre naif de Bayes.

```python
import numpy as np

def map_naif(M):
    N = len(M)
    n_G = np.sum(M[:, 2] == 'G')
    n_P = np.sum(M[:, 2] == 'P')

    P_G = n_G / N
    P_P = n_P / N
    P_BT = np.sum((M[:, 0] == 'B') & (M[:, 1] == 'T')) / N

    P_B_sachant_G = np.sum((M[:, 0] == 'B') & (M[:, 2] == 'G')) / n_G
    P_T_sachant_G = np.sum((M[:, 1] == 'T') & (M[:, 2] == 'G')) / n_G
    P_B_sachant_P = np.sum((M[:, 0] == 'B') & (M[:, 2] == 'P')) / n_P
    P_T_sachant_P = np.sum((M[:, 1] == 'T') & (M[:, 2] == 'P')) / n_P

    P_BT_sachant_G = P_B_sachant_G * P_T_sachant_G
    P_BT_sachant_P = P_B_sachant_P * P_T_sachant_P

    P_G_sachant_BT = P_BT_sachant_G * P_G / P_BT
    P_P_sachant_BT = P_BT_sachant_P * P_P / P_BT

    print(f"P(G) = {P_G:.4f}   P(P) = {P_P:.4f}   P(B,T) = {P_BT:.4f}")
    print(f"P(B,T|G) = {P_BT_sachant_G:.4f}   P(B,T|P) = {P_BT_sachant_P:.4f}")
    print(f"P(G|B,T) = {P_G_sachant_BT:.4f}   P(P|B,T) = {P_P_sachant_BT:.4f}")

    if P_G_sachant_BT > P_P_sachant_BT:
        return 'G'
    elif P_G_sachant_BT < P_P_sachant_BT:
        return 'P'
    else:
        return 'égalité'
```

```python
map_naif(A)
```

# D – Maximum de vraisemblance

1) Solution de maximum de vraisemblance en utilisant le cadre naif de Bayes.

```python
def mle_naif(M):
    n_G = np.sum(M[:, 2] == 'G')
    n_P = np.sum(M[:, 2] == 'P')

    P_B_sachant_G = np.sum((M[:, 0] == 'B') & (M[:, 2] == 'G')) / n_G
    P_T_sachant_G = np.sum((M[:, 1] == 'T') & (M[:, 2] == 'G')) / n_G
    P_B_sachant_P = np.sum((M[:, 0] == 'B') & (M[:, 2] == 'P')) / n_P
    P_T_sachant_P = np.sum((M[:, 1] == 'T') & (M[:, 2] == 'P')) / n_P

    P_BT_sachant_G = P_B_sachant_G * P_T_sachant_G
    P_BT_sachant_P = P_B_sachant_P * P_T_sachant_P

    print(f"P(B,T|G) = {P_BT_sachant_G:.4f}   P(B,T|P) = {P_BT_sachant_P:.4f}")

    if P_BT_sachant_G > P_BT_sachant_P:
        return 'G'
    elif P_BT_sachant_G < P_BT_sachant_P:
        return 'P'
    else:
        return 'égalité'
```
