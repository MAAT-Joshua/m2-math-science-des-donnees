```python
import numpy as np
import matplotlib.pyplot as plt    
import pandas as pd
from sklearn import datasets
from sklearn import metrics
from sklearn.metrics.pairwise import euclidean_distances
from sklearn.neighbors import KNeighborsClassifier

iris = datasets.load_iris()
X = iris.data
Y = iris.target
```

```python
metrics.pairwise.euclidean_distances(X)
```

```python
def PPV(X,Y) :
    i=0
    Z = []
    c = metrics.pairwise.euclidean_distances(X)
    for i in range(len(X)):
        a = np.argsort(c[i])
        b = a[1]
        Z.append(Y[b])
    return Z
```

```python
Y_2 = PPV(X,Y)
```

2) on veut vérifier le taux d'erreur

```python
Y_3 = Y - Y_2
err = f"{np.count_nonzero(Y_3)/len(Y_3) : .0%}"
print(err)
```

# Bonus 
On veut maintenant que notre fonction PPV puisse faire pour k voisins :

```python
def PPV(X,Y,K) :
    i=0
    Z = []
    c = metrics.pairwise.euclidean_distances(X)
    for i in range(len(X)):
        a = np.argsort(c[i])
        a = pd.Series(Y[a[1:K+1]])
        b = a.mode()
        if len(b)>1:
            return PPV(X,Y,K+1)
        b = b.item()
        Z.append(b)
    return Z
```

On a comparé chaque étiquette des plus proche voisin et en cas d'égalité j'ai arbitré en lançant sur K+1 voisin pour trancher

```python
PPV(X,Y,1)
```

# Classifieur Bayesien Naïf

1) On va d'abord trouve les barycentres de nos classes

```python
def barycentre(X):
    b = np.mean(X, axis = 0)
    return b
```

On calcul alors le barycentre de chaque classe

```python
def barycentre_C(X,Y):
    c = []
    u = np.unique(Y)
    for i in u :
        p = X[Y == i]
        c.append(barycentre(p))
    np.asarray(c)
    return c
```

On calcule les probabilité apriori d'appartenir à une classe :

```python
def proba_classe(Y):
    c = []
    u = np.unique(Y)
    for classe in u :
        n = (Y == classe).sum()/len(Y)
        c.append(n)
    return c
```

```python
proba_classe(Y)
```

On s'occupe alors des proba conditionnelles de chaque donnée

```python
def dist_point_baryclasse(X,Y) :
    u = np.unique(Y)
    predict = []
    bary = barycentre_C(X,Y)
    for i in range(len(X)):
        distbary = []
        for classe in u :
            dist = np.linalg.norm(X[i] - bary[classe])
            distbary.append(dist)
        indice = np.argmin(distbary)
        predict.append(u[indice])
    return predict
```

```python
dist_point_baryclasse(X,Y)
```

On implémente maintenant la fonction CBN complète

```python
def CBN(X, Y):
    u = np.unique(Y)
    predict = []
    bary = barycentre_C(X, Y)
    probas_prior = proba_classe(Y)

    for i in range(len(X)):
        prob_classes = []
        for idx_classe in range(len(u)):
            p_xi_wk_produit = 1.0
            for j in range(len(X[i])):
                valeur_donnee = X[i][j]
                dist_classe_actuelle = abs(valeur_donnee - bary[idx_classe][j])
                somme_distances = 0
                for c_idx in range(len(u)):
                    somme_distances += abs(valeur_donnee - bary[c_idx][j])
                if somme_distances == 0:
                    p_variable = 1.0 / len(u)
                else:
                    p_variable = 1.0 - (dist_classe_actuelle / somme_distances)
                p_xi_wk_produit *= p_variable
            prob_posteriore = p_xi_wk_produit * probas_prior[idx_classe]
            prob_classes.append(prob_posteriore)
        indice_max = np.argmax(prob_classes)
        predict.append(u[indice_max])

    return predict
```

```python
Y_2 = CBN(X,Y)
```

```python
Y_8 = Y - Y_2
err = f"{np.count_nonzero(Y_8)/len(Y_8) : .0%}"
print(err)
```
