# D. Génération de données et affichage

Pour répondre à la question concernant nos données : dans les arguments de la fonction datasets.make_blobs, random_state permet de fixer des données, même si elles sont tirées de manière aléatoire ça permet de les fixer une fois tirée pour réutiliser ces mêmes données aléatoire plutot qu'en tirrer de nouvelles a chaque fois que l'on exécute le code.

```python
help
```

```python
help(datasets.make_blobs)
```

```python
test = make_blobs(n_samples=1000, n_features=2, centers=4, cluster_std=1.0, center_box=(-10.0, 10.0), shuffle=True, random_state=None, return_centers=False)
print(test)
```

# On affiche alors notre figure

```python
plt.figure(figsize=(10, 10))
```

```python
X_data = test[0]
```

```python
x_coords = X_data[:, 0]
y_coords = X_data[:, 1] 
```

```python
help(plt.scatter)
```

```python
plt.scatter(x_coords, y_coords,c = test[1])
plt.title("Visualisation clusters")
plt.xlim([-15, 15])
plt.ylim([-15, 15])
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

# D)4) Deuxième tableau et concat

```python
test2 = make_blobs(n_samples=100, n_features=2, centers=2, cluster_std=1.0, center_box=(-10.0, 10.0), shuffle=True, random_state=None, return_centers=False)
test3 = make_blobs(n_samples=500, n_features=2, centers=3, cluster_std=1.0, center_box=(-10.0, 10.0), shuffle=True, random_state=None, return_centers=False)
```

```python
help(hstack)
```

```python
vstack((test2[0],test3[0]))
hstack((test2[1],test3[1]))
```

# Affichage des 3 ensembles avec deux concaténés

```python
X_23 = vstack((test2[0],test3[0]))
Y_23 = hstack((test2[1],test3[1]))


plt.figure(figsize=(10, 10))


plt.scatter(x_coords, y_coords,c = test[1])
plt.scatter(X_23[:,0], X_23[:,1],c = Y_23)

plt.title("Visualisation clusters")
plt.xlim([-15, 15])
plt.ylim([-15, 15])
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
